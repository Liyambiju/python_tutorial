## caesar_cipher

```python
text = input("Enter text: ")
shift = int(input("Enter shift value: "))

encrypted = ""
decrypted = ""

# Encryption
for ch in text:
    if ch.isalpha():
        encrypted += chr((ord(ch) - 97 + shift) % 26 + 97)
    else:
        encrypted += ch

# Decryption
for ch in encrypted:
    if ch.isalpha():
        decrypted += chr((ord(ch) - 97 - shift) % 26 + 97)
    else:
        decrypted += ch

print("Encrypted text:", encrypted)
print("Decrypted text:", decrypted)
```
